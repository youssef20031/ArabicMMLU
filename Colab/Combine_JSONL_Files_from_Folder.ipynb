```
import os
import glob # Import glob for finding files matching a pattern

def combine_jsonl_files_from_folder(input_folder, output_file):
    """
    Reads all JSONL files from a specified folder and writes their contents
    sequentially to a single output JSONL file.

    Args:
        input_folder (str): The path to the folder containing input JSONL files.
        output_file (str): The path to the output JSONL file.
    """
    print(f"Starting the combination process...")
    print(f"Input folder: {input_folder}")
    print(f"Output file: {output_file}")

    # Check if the input folder exists
    if not os.path.isdir(input_folder):
        print(f"Error: Input folder not found: {input_folder}")
        print("Please ensure the folder exists and the path is correct.")
        return # Exit if the folder doesn't exist

    # Find all files ending with .jsonl in the specified folder
    # os.path.join ensures correct path construction (e.g., data/*.jsonl)
    # glob.glob finds all pathnames matching the specified pattern
    # Using recursive=False to only search the top-level directory
    input_files = glob.glob(os.path.join(input_folder, '*.jsonl'))

    if not input_files:
        print(f"No .jsonl files found directly in folder: {input_folder}")
        # Optional: Check subdirectories if needed
        # input_files = glob.glob(os.path.join(input_folder, '**/*.jsonl'), recursive=True)
        # if not input_files:
        #     print(f"No .jsonl files found in folder or subfolders: {input_folder}")
        #     return
        return # Exit if no jsonl files are found

    # Sort files alphabetically to ensure consistent order (optional but good practice)
    input_files.sort()

    print(f"Found {len(input_files)} .jsonl files to combine:")
    for f in input_files:
        # Use os.path.relpath to show path relative to input_folder for clarity
        print(f" - {os.path.relpath(f, input_folder)}")

    try:
        # Open the output file in write mode ('w') with utf-8 encoding
        # Using 'with' ensures the file is automatically closed even if errors occur
        with open(output_file, 'w', encoding='utf-8') as outfile:
            # Iterate through each found input file path
            for infile_path in input_files:
                try:
                    # Open the current input file in read mode ('r') with utf-8 encoding
                    with open(infile_path, 'r', encoding='utf-8') as infile:
                        print(f"Processing file: {os.path.basename(infile_path)}...")
                        line_count = 0
                        # Read each line from the current input file
                        for line in infile:
                            # Basic check for non-empty lines before writing
                            if line.strip():
                                # Write the line directly to the output file
                                outfile.write(line)
                                # Ensure each line ends with a newline, crucial for JSONL
                                if not line.endswith('\n'):
                                    outfile.write('\n')
                                line_count += 1
                            else:
                                print(f"  - Skipped empty line in {os.path.basename(infile_path)}")

                    print(f"Finished processing: {os.path.basename(infile_path)} ({line_count} lines added)")

                except Exception as e:
                    print(f"Error reading file {infile_path}: {e}. Skipping.")
                    continue # Skip to the next file if an error occurs during reading

        print(f"\nSuccessfully combined {len(input_files)} files into: {output_file}")

    except IOError as e:
        print(f"Error opening or writing to output file {output_file}: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    # --- Configuration for Colab ---
    # 1. Set the path to the folder containing your .jsonl files.
    #    If you created a folder named 'data' in your Colab environment's
    #    root directory and uploaded your files there, use '/content/data'.
    #    Use the "Files" tab on the left in Colab to check the exact path.
    input_folder_path = "/content/data"  # <-- CHANGE THIS if your folder has a different name/location

    # 2. Set the desired name for the combined output file.
    #    This file will be created in your Colab environment's root directory ('/content/').
    output_file_path = "/content/combined_output.jsonl" # <-- CHANGE THIS if you want a different output name/location
    # --- End Configuration ---

    # Create the input directory if it doesn't exist (useful for initial setup)
    if not os.path.exists(input_folder_path):
        print(f"Input folder '{input_folder_path}' not found. Creating it.")
        print("Please upload your .jsonl files to this folder using the Colab file browser.")
        os.makedirs(input_folder_path)
        # We stop here if the folder was just created, as there are no files yet.
    else:
        # Call the function to combine the files from the folder
        combine_jsonl_files_from_folder(input_folder_path, output_file_path)

    # --- How to Run in Colab ---
    # 1. Make sure this script is in a Colab code cell.
    # 2. Use the "Files" tab (folder icon on the left sidebar) in Colab:
    #    - Create a folder (e.g., right-click -> New folder -> name it 'data').
    #    - Upload your .jsonl files into that 'data' folder.
    # 3. Modify the `input_folder_path` variable above if your folder isn't exactly '/content/data'.
    # 4. Modify the `output_file_path` variable if you want a different name or location for the output.
    # 5. Run this code cell (click the play button or use Shift+Enter).
    # 6. The combined file (e.g., 'combined_output.jsonl') will appear in the location specified by `output_file_path` (usually the root '/content/' directory). You can then download it from the "Files" tab.
```
