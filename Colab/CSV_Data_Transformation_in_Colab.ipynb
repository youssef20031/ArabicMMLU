```
import csv

def process_csv(input_file_path, output_file_path):
    """
    Reads a CSV file, copies Arabic fields to standard fields, and writes to a new CSV.

    Args:
        input_file_path (str): The path to the input CSV file.
        output_file_path (str): The path where the new CSV file will be saved.
    """
    try:
        with open(input_file_path, 'r', newline='', encoding='utf-8') as infile, \
             open(output_file_path, 'w', newline='', encoding='utf-8') as outfile:

            reader = csv.DictReader(infile)
            # Define the fieldnames for the output file
            # These are the new column headers
            fieldnames_output = ['observation_1', 'observation_2', 'hypothesis_1', 'hypothesis_2', 'label']
            writer = csv.DictWriter(outfile, fieldnames=fieldnames_output)

            # Write the header row to the output file
            writer.writeheader()

            # Process each row from the input file
            for row in reader:
                # Create a new dictionary for the output row
                # This dictionary will map the new column names to the correct values from the input row
                output_row = {
                    'observation_1': row.get('observation_1_ar', ''), # Get Arabic observation 1, default to empty string if not found
                    'observation_2': row.get('observation_2_ar', ''), # Get Arabic observation 2, default to empty string if not found
                    'hypothesis_1': row.get('hypothesis_1_ar', ''),  # Get Arabic hypothesis 1, default to empty string if not found
                    'hypothesis_2': row.get('hypothesis_2_ar', ''),  # Get Arabic hypothesis 2, default to empty string if not found
                    'label': row.get('label', '')                     # Copy the label as is, default to empty string if not found
                }
                writer.writerow(output_row)
        print(f"Processing complete. Output file saved to: {output_file_path}")

    except FileNotFoundError:
        print(f"Error: The file '{input_file_path}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- How to use in Google Colab ---

# 1. Upload your CSV file:
#    In Colab, click on the folder icon on the left sidebar.
#    Click the "Upload to session storage" button (upward arrow icon) and select your CSV file.
#    Let's assume your uploaded file is named 'input_data.csv'.

# 2. Define the input and output file paths:
input_csv_file = 'input_data.csv'  # Replace with the actual name of your uploaded file
output_csv_file = 'output_data_ar.csv' # Name for your new processed file

# 3. Run the processing function:
process_csv(input_csv_file, output_csv_file)

# 4. Download the new file:
#    After the script runs, the 'output_data_ar.csv' file will appear in the Colab file explorer.
#    You can right-click on it and select "Download" to save it to your computer.

# --- Example of how the input CSV structure is expected ---
# observation_1,observation_2,hypothesis_1,hypothesis_2,label,observation_1_ar,observation_2_ar,hypothesis_1_ar,hypothesis_2_ar
# "English obs1","English obs2","English hyp1","English hyp2","positive","ملاحظة عربية 1","ملاحظة عربية 2","فرضية عربية 1","فرضية عربية 2"
# "Another Eng obs1","Another Eng obs2","Another Eng hyp1","Another Eng hyp2","negative","ملاحظة أخرى 1","ملاحظة أخرى 2","فرضية أخرى 1","فرضية أخرى 2"

# --- Example of how the output CSV structure will be ---
# observation_1,observation_2,hypothesis_1,hypothesis_2,label
# "ملاحظة عربية 1","ملاحظة عربية 2","فرضية عربية 1","فرضية عربية 2","positive"
# "ملاحظة أخرى 1","ملاحظة أخرى 2","فرضية أخرى 1","فرضية أخرى 2","negative"
```
