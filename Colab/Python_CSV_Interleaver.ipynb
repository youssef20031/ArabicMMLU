```
import csv
import math

def interleave_csv(input_filename, output_filename, num_output_data_lines):
    """
    Reads a CSV file, and selects a specified number of data lines in an interleaved manner.

    Args:
        input_filename (str): The name of the input CSV file.
        output_filename (str): The name for the output CSV file.
        num_output_data_lines (int): The number of data lines to select for the output.
    """
    try:
        with open(input_filename, 'r', newline='', encoding='utf-8') as infile:
            reader = csv.reader(infile)
            all_rows = list(reader)
    except FileNotFoundError:
        print(f"Error: Input file '{input_filename}' not found. Please make sure the file exists in the same directory as the script, or provide the full path.")
        return
    except Exception as e:
        print(f"Error reading CSV file '{input_filename}': {e}")
        return

    if not all_rows:
        print(f"Error: Input CSV file '{input_filename}' is empty.")
        return

    # Assume the first row is the header
    header = all_rows[0]
    data_rows = all_rows[1:]

    num_input_data_rows = len(data_rows)

    if num_input_data_rows == 0:
        print(f"Warning: Input CSV file '{input_filename}' has a header but no data rows.")
        # Optionally, write only the header to the output file
        try:
            with open(output_filename, 'w', newline='', encoding='utf-8') as outfile:
                writer = csv.writer(outfile)
                writer.writerow(header)
            print(f"Output file '{output_filename}' created with header only.")
        except Exception as e:
            print(f"Error writing output CSV file '{output_filename}': {e}")
        return

    selected_data_rows = []

    if num_input_data_rows < num_output_data_lines:
        print(f"Warning: Input file has only {num_input_data_rows} data rows, which is less than the requested {num_output_data_lines} lines.")
        print(f"Selecting all {num_input_data_rows} available data rows instead.")
        selected_data_rows = data_rows
    elif num_output_data_lines <= 0:
        print(f"Info: Requested number of output lines is {num_output_data_lines}. No data lines will be selected beyond the header.")
        # selected_data_rows remains empty
    else:
        # Calculate the step to pick lines at regular intervals
        # step = total available data lines / number of lines we want to select
        step = num_input_data_rows / num_output_data_lines

        for i in range(num_output_data_lines):
            # Calculate the index of the line to pick from the original data_rows
            # round() is used to get the nearest whole number index
            original_index = round(i * step)

            # Ensure the calculated index is within the bounds of available data_rows
            # This is a safeguard, especially for the last few selections or if num_input_data_rows is small
            actual_index = min(original_index, num_input_data_rows - 1)

            selected_data_rows.append(data_rows[actual_index])

        # Note: For step > 1 (num_input_data_rows > num_output_data_lines),
        # the `round(i * step)` method should naturally produce unique, ordered indices.

    try:
        with open(output_filename, 'w', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            writer.writerow(header)  # Write the header first
            writer.writerows(selected_data_rows) # Write the selected data rows

        print(f"Successfully created '{output_filename}' with the header and {len(selected_data_rows)} interleaved data lines.")
        if num_input_data_rows >= num_output_data_lines and num_output_data_lines > 0 :
             print(f"Selected approximately every {num_input_data_rows / num_output_data_lines:.2f}-th data line from the input file.")

    except Exception as e:
        print(f"Error writing output CSV file '{output_filename}': {e}")

if __name__ == "__main__":
    # --- Configuration for your specific request ---
    # Your input CSV file is expected to have 7377 lines total (1 header + 7376 data lines)
    # We want 1000 data lines in the output CSV.

    input_csv_file = "training_deductive.csv"  # <<<< CHANGE THIS if your input file has a different name
    output_csv_file = "output_interleaved.csv" # <<<< CHANGE THIS if you want a different output name
    number_of_data_lines_to_output = 1000

    print(f"Starting CSV interleaving process...")
    print(f"Input file: {input_csv_file}")
    print(f"Output file: {output_csv_file}")
    print(f"Number of data lines to select: {number_of_data_lines_to_output}")

    # Call the function with your specific parameters
    interleave_csv(input_csv_file, output_csv_file, number_of_data_lines_to_output)
```
